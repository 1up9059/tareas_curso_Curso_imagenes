## Show image

```python
from matplotlib import pyplot as plt

def show_image(image_tensor, title = "Image", vmax = 1):
  plt.rcParams["axes.grid"] = False
  plt.figure()
  plt.imshow(image_tensor, cmap=plt.get_cmap('gray'),vmin=0,vmax = vmax)
  plt.title(title)
```

## Open image

```python
from __future__ import print_function
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
import cv2
import sys

"""
Open image
@param path, path to open image
@return imageTensor, pytorch tensor matrix
"""
def open_image(path):
  #use cv2 to open image
  image = cv2.imread(path, 0) # 0 is grayscale
  #convert to pytorch tensor
  imageTensor = torch.FloatTensor(image)
  return imageTensor, image
```

## Morphological filers

### Erosion

```python
"""
Perform the convolution
@param U, input image
@param F, filter
@return Y, filtered image
"""
def erosion(U, F):
    #Dimensions of input image
    dimensionsInputImage = U.size()
    #sum of F values
    sum_F = F.sum()
    #output image
    Y = torch.zeros(dimensionsInputImage)
    
    #Filter dimensions
    dimensionsFilter = F.size()
    
    # Filter size
    filterSize = dimensionsFilter[0]
    
    # Calcule the padding (half padding)
    padd_size = filterSize // 2
    
    # Add the padding to the image
    padding = nn.ConstantPad2d((padd_size, padd_size, padd_size, padd_size), 0)
    U = padding(U)
    # convolutions
    for i in range(0, dimensionsInputImage[0]):
        for j in range(0, dimensionsInputImage[1]):
            windowU = U[i: filterSize + i, j: filterSize + j]
            #dp            
            elementWiseProduct = torch.sum(windowU * F)
            #print("sum_F ", sum_F)
            #print("elementWiseProduct ", elementWiseProduct)
            if(elementWiseProduct == sum_F):
               #print("assigned")
               Y[i, j] = 1
            
            
    return Y
```

```python
def test_erosion():
  U, image = open_image("binary1.png")
  #binarize the image to make sure is just 0s and 1s
  U[U > 0] = 1
  show_image(U, "Input image")
  S = torch.ones(5, 5)
  Y = erosion(U, S)
  show_image(Y, "Erosion")
test_erosion()
```
